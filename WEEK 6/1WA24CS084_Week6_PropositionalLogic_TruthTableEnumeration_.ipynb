```python
from itertools import product
from tabulate import tabulate

def implies(a, b):
    return (not a) or b

def check_kb(P, Q, R):
    s1 = implies(Q, P)
    s2 = implies(P, not Q)
    s3 = Q or R

    kb = s1 and s2 and s3

    return s1, s2, s3, kb


def tt_entails(query_function):
    for P, Q, R in product([False, True], repeat=3):

        s1, s2, s3, kb = check_kb(P, Q, R)

        if kb and not query_function(P, Q, R):
            return False

    return True


# Generate complete truth table
headers = ["P", "Q", "R", "Q -> P", "P -> ~Q", "Q | R", "KB"]

table = []

for P, Q, R in product([False, True], repeat=3):

    s1, s2, s3, kb = check_kb(P, Q, R)

    table.append([
        P, Q, R, s1, s2, s3, kb
    ])

print("\nCOMPLETE TRUTH TABLE")
print(tabulate(table, headers=headers, tablefmt="grid"))


# Check queries
q1 = tt_entails(lambda P, Q, R: R)
q2 = tt_entails(lambda P, Q, R: implies(R, P))
q3 = tt_entails(lambda P, Q, R: implies(Q, R))


# Display entailment results
results = [
    ["KB |= R", q1],
    ["KB |= R -> P", q2],
    ["KB |= Q -> R", q3]
]

print("\nENTAILMENT RESULTS")
print(tabulate(results, headers=["Query", "Entailment"], tablefmt="grid"))
```
